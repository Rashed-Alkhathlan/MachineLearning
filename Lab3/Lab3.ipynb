```python
# Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Make plots look cleaner
sns.set_theme()
```

```python
# Load Dataset

df = pd.read_csv("amazon.csv")

# Display first 5 rows
df.head()
```

### Check Missing Values

```python
# Check all values
print(df.isna())
```

```python
print(df.isna().sum())
```

### Check duplicate rows

```python
# checking duplicate rows
df[df.duplicated()]
```

### No. of rows and columns

```python
# finding number of rows and columns

print("Shape (rows, columns): ", df.shape,"\n")

print("number of rows: ", df.shape[0])
print("number of columns: ", df.shape[1])
```

### Data type of columns

```python
# viewing the data types of columns
df.dtypes
```

### Cleaning data

```python
df['discounted_price'] = df['discounted_price'].replace(r'[\₹,]', '', regex=True)
df['actual_price'] = df['actual_price'].replace(r'[\₹,]', '', regex=True)
df['rating_count'] = df['rating_count'].replace(r'[\,]', '', regex=True)
df['discount_percentage'] = df['discount_percentage'].replace(r'[\%]', '', regex=True)

cols = ['discounted_price', 'actual_price', 'discount_percentage', 'rating', 'rating_count']

# Convert float columns
for col in cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')  # invalid parsing becomes NaN

df.dtypes
```

```python
# Split the pipe-separated categories into lists
df['category_list'] = df['category'].apply(lambda x: str(x).split('|'))
df[['category', 'category_list']].head()
```

### Descriptive summary Statistics

```python
# Statistical summary
df.describe(include='all')
```

### Univariate Analysis

```python
numeric_cols = ['discounted_price','actual_price','discount_percentage','rating','rating_count']
for col in numeric_cols:
    plt.figure(figsize=(8,4))
    sns.histplot(df[col], bins=20, kde=True)
    plt.title(f'Distribution of {col}')
    plt.show()
```

## Bivariate Analysis

### Numerical vs Numerical

```python
numeric_cols = ['discounted_price','actual_price','discount_percentage','rating','rating_count']
corr = df[numeric_cols].corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

```python
sns.scatterplot(x='discount_percentage', y='rating', data=df)
plt.title('Rating vs Discounted Percentage')
plt.show()
```

### Categorical vs Numerical

```python
# Create a row for each category per product
df_exploded = df.explode('category_list')
df_exploded = df_exploded.reset_index(drop=True)
df_exploded['category_list'] = df_exploded['category_list'].str.strip()
df_exploded['category_list'] = df_exploded['category_list'].str.replace('\n','')  # remove newlines if any
df_exploded['category_list'] = df_exploded['category_list'].str.replace('\t','')  # remove tabs
df_exploded['category_list'] = df_exploded['category_list'].str.lower()  # optional: lowercase everything

df_exploded.head()
```

```python
top_categories = df_exploded['category_list'].value_counts().head(10).reset_index()

sns.barplot(x=top_categories['category_list'], y=top_categories['count'])
plt.xticks(rotation=60)
plt.title('Top 10 Most Frequent Categories')
plt.show()
```

```python
df_top = df_exploded[df_exploded['category_list'].isin(top_categories['category_list'])]

plt.figure(figsize=(12,6))
sns.boxplot(x='category_list', y='rating', data=df_top)
plt.xticks(rotation=45)
plt.title('Rating Distribution per Category')
plt.show()
```
